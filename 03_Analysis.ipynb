```python
analyser = SentimentIntensityAnalyzer()
```

```python
import seaborn as sns, numpy as np
```

```python
##Sentiment Analysis

## POSITIVE: 1 // NEUTRAL: -0.25 to +0.25 // NEGATIVE: Below -0.25
def sentiment_analyzer_scores(text, engl=True):

    trans = text
    #print(trans)

    score = analyser.polarity_scores(trans)
    lb = score['compound']
    print(lb)
    if lb >= 0.05:
        return 1
    elif (lb > -0.05) and (lb < 0.05):
        return 0
    elif (lb<=-0.05):
        return -1
```

```python

def anl_tweets(lst, title='Tweets Sentiment', engl=True ):
    sents = []
    #i=-1
    for tw in lst:
        #i += 1
        try:
            st = sentiment_analyzer_scores(tw['datos'], engl)
            sents.append(st)

        except:
            sents.append(0)

    ax = sns.distplot(
        sents,
        kde=False,
        bins=3)
    ax.set(xlabel='Negative                Neutral                 Positive',
           ylabel='#Tweets',
          title="Tweets of @"+title)

    return sents
```

```python
anl_tweets(tw_dataset, 'Retirement', True)
```

```python

def sentiment_analyzer_scores_pos(lst, engl=True):

    sents=[]
    for tw in lst:

        score = analyser.polarity_scores(tw['datos'])
        lb = score['compound']
        if lb >=0.05:
            sents.append(tw) ##texto positivo.

    return sents
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from os import path
from PIL import Image
from wordcloud import WordCloud, STOPWORDS, ImageColorGenerator
```

```python
#word_cloud

def word_cloud(wd_list):
    stopwords = set(STOPWORDS)
    all_words = ' '.join([text['datos'] for text in wd_list])
    ##print(all_words)
    wordcloud = WordCloud(
        background_color='white',
        stopwords=stopwords2,
        width=1600,
        height=800,
        random_state=21,
        colormap='seismic', #'jet', #'rainbow', #'hot', #'jet',
        max_words=50,
        max_font_size=500).generate(all_words)
    plt.figure(figsize=(12, 10))
    plt.axis('off')
    plt.imshow(wordcloud, interpolation="bilinear");
```
